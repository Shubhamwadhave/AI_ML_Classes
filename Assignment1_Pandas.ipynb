```python
#Q25.	Convert the hours logged column to a number.
df['hours_logged'] = pd.to_numeric(df['hours_logged'], errors='coerce')

# The hours logged column is converted from text into number
```

```python

#Q26.	Convert the billing rate column to a number. Remove the $ sign first.

df['billing_rate_(usd)'] = df['billing_rate_(usd)'].astype(str).str.replace('$', '', regex=False)
df['billing_rate_(usd)'] = pd.to_numeric(df['billing_rate_(usd)'], errors='coerce')

# The billing rate is cleaned by removing the dollar or USD text and then converted into numbers
```

```python
#Q27.	Convert the CSAT score column to a number.
df["csat_score"] = pd.to_numeric(
    df["csat_score"].astype(str).str.extract(r"([-+]?\d*\.?\d+)")[0],
    errors="coerce"
)


df["csat_score"].dtype

# The CSAT score is converted into numeric values so we can calculate its average and other values
```

```python
#Q28.	Print the average, the smallest and the largest value of hours logged.
print("Average:", df["hours_logged"].mean())
print("Minimum:", df["hours_logged"].min())
print("Maximum:", df["hours_logged"].max())

 # This gives the average, minimum and maximum hours logged in the tickets
```

```python
#Q29.	Sort the table by hours logged from highest to lowest and show the top 10 rows.
df.sort_values("hours_logged", ascending=False).head(10)

# This shows the 10 tickets which have the highest logged hours
```

```python
#Q30.	Show the 5 tickets with the highest billable amount.

# Q30
df["billable_amount_usd"] = (
    df["billable_amount_usd"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace("USD", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["billable_amount_usd"] = pd.to_numeric(
    df["billable_amount_usd"],
    errors="coerce"
)

df.nlargest(5, "billable_amount_usd")

# This shows the five tickets having the highest billable amount
```

```python
#Q31.	Remove the extra spaces from the client name column.
df["client_name"] = df["client_name"].str.strip()

# Extra spaces from the client names are removed
```

```python
#Q32.	Change the client name column to title case.
df["client_name"] = df["client_name"].str.title()

# The client names are changed into title case
```

```python
#Q33.	Convert the ticket created column to a date.
df["ticket_created_at"] = pd.to_datetime(
    df["ticket_created_at"],
    errors="coerce",
    format="mixed"
)

df["ticket_created_at"].dtype

# The ticket created column is converted into a proper date format.
```

```python
#Q34.	Add a new column that holds the year of the ticket.

df["year"] = df["ticket_created_at"].dt.year

df[["ticket_created_at", "year"]].head()

# A new year column is added from the ticket creation date.
```

```python
#Q35.	Add a new column that holds the month of the ticket.
df["month"] = df["ticket_created_at"].dt.month

df[["ticket_created_at", "month"]].head()

# A new month column is added to show which month the ticket was created.
```

```python
#Q36.	Count how many tickets were created in each year.
df["year"].value_counts().sort_index()

# This shows how many tickets were created in each year.
```

```python
#Q37.	Add a new column called total equal to billing rate multiplied by hours logged.
df["total"] = df["billing_rate_(usd)"] * df["hours_logged"]

df[["billing_rate_(usd)", "hours_logged", "total"]].head()

# The new total column is calculated by multiplying billing rate with hours logged.
```

```python
#Q38.	Fill the missing values in the CSAT score column with the average CSAT score.
average_csat = df["csat_score"].mean()

df["csat_score"] = df["csat_score"].fillna(average_csat)

df["csat_score"].isna().sum()

# Missing CSAT values are filled using the average CSAT score.
```

```python
#Q39.	Drop the rows where the client name is missing.
df = df.dropna(subset=["client_name"])

print("Rows:", len(df))

# Rows where the client name is missing are removed from the dataset.
```

```python
#Q40.	Group by service line and find the average hours logged.
df.groupby("serviceline")["hours_logged"].mean()

# This shows the average hours logged for each service line.
```

```python
#Q41.	Group by priority and count the tickets.
df.groupby("priority").size()

#This gives the number of tickets available for each priority.
```

```python
#Q42.	Group by client name and find the total billable amount.
df.groupby("client_name")["billable_amount_usd"].sum().sort_values(ascending=False)

# This shows the total billable amount for each client.
```

```python
#Q43.	Find the top 5 clients by number of tickets.
df["client_name"].value_counts().head(5)

# This gives the five clients which have the highest number of tickets.
```

```python
#Q44.	Find the average CSAT score for each priority.
df.groupby("priority")["csat_score"].mean()

# This shows the average CSAT score for each priority level.
```

```python
#Q45.	Draw a bar chart of ticket count by service line.
df["serviceline"].value_counts().plot(
    kind="bar",
    figsize=(10, 5),
    title="Ticket Count by Service Line"
)

import matplotlib.pyplot as plt
plt.xlabel("Service Line")
plt.ylabel("Number of Tickets")
plt.show()

# The bar chart makes it easier to compare the number of tickets between different service lines.
```

```python
#Q46.	Draw a histogram of hours logged.
df["hours_logged"].plot(
    kind="hist",
    bins=30,
    figsize=(8, 5),
    title="Distribution of Hours Logged"
)

import matplotlib.pyplot as plt
plt.xlabel("Hours Logged")
plt.show()

# The histogram shows how the logged hours are spread across the tickets.
```

```python
#Q47.	Show only the tickets where hours logged is more than 50.
df[df["hours_logged"] > 50]

# This shows all the tickets where more than 50 hours were logged.
```

```python
#Q48.	Show the tickets where status is Closed and priority is P1.
df[
    (df["status"].astype(str).str.strip().str.lower() == "closed") &
    (df["priority"].astype(str).str.strip().str.upper() == "P1")
]

# This shows tickets which are both closed and have P1 priority.
```

```python
#Q49.	In the SLA breached column, change Yes, Y, TRUE and 1 to True, and everything else to False.
df["sla_breached"] = (
    df["sla_breached"]
    .astype(str)
    .str.strip()
    .str.lower()
    .isin(["yes", "y", "true", "1"])
)

df["sla_breached"].value_counts()


# The SLA breached values are converted into True or False based on the given conditions.
```

```python
#Q50.	Save your cleaned table to a new CSV file.
df.to_csv("it_services_tickets_cleaned.csv", index=False)

print("Cleaned file saved successfully.")

# The cleaned DataFrame is saved as a new CSV file for future use.
```
